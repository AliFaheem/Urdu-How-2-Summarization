```python
import subprocess
import time

# Step 1: Start Ollama service
def start_ollama_service():
    # Start the Ollama server
    process = subprocess.Popen(["ollama", "serve"], stdout=subprocess.PIPE, stderr=subprocess.PIPE)
    return process

# Step 2: Run a model (like llama3)
def run_model(model_name="llama3.1"):
    # Make sure the Ollama server is running before trying to run the model
    time.sleep(5)  # Wait a few seconds for the server to start
    result = subprocess.run(["ollama", "run", model_name], capture_output=True, text=True)
    return result.stdout

if __name__ == "__main__":
    # Start Ollama service
    process = start_ollama_service()
    print("Starting Ollama service...")

    # Wait for the server to initialize
    time.sleep(3)  # Add some wait time before running the model
    
    # Run the model
    output = run_model("llama3")
    print("Model Output:")
    print(output)
    
    # Optionally, stop the Ollama server after use
    process.terminate()
    print("Ollama service stopped.")
```

```python
import time
from ollama import Client

# Create the client once
client = Client(
    host='http://localhost:11434',
    headers={'x-some-header': 'some-value'}
)

# Function to measure the time taken for multiple requests
def measure_execution_time():
    start_time = time.time()  # Start the timer

    # Make repeated chat calls (10 times)
    for _ in range(10):
        response = client.chat(model='llama3', messages=[
            {'role': 'user', 'content': 'Why is the sky blue?'}
        ])
        
        # Print the response content (or process the response further)
        # print(response['message']['content'])

    end_time = time.time()  # End the timer

    # Calculate and print the total time taken
    total_time = end_time - start_time
    print(f"Total time taken for 10 requests: {total_time:.4f} seconds")
    print(f"Average time per request: {total_time / 10:.4f} seconds")

# Call the function to measure execution time
measure_execution_time()
```
